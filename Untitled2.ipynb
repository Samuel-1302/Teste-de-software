```python
%%writefile cofrinho.py

def adicionar_deposito(cofre, valor):
    """Adiciona um depósito ao cofre e retorna o cofre atualizado."""
    cofre.append(valor)
    return cofre

def calcular_total_economizado(cofre):
    """Retorna a soma de todos os depósitos feitos."""
    total = 0
    for deposito in cofre:
        total += deposito
    return round(total, 2)

def meta_atingida(total, meta):
    """Retorna True se o total economizado atingiu a meta definida."""
    return total >= meta
```

```python
%%writefile test_cofrinho.py
from cofrinho import adicionar_deposito, calcular_total_economizado, meta_atingida
def test_adicionar_deposito_ao_cofre_vazio():
  # Arrange (Organizar dados, iniciar variáveis, ...)
    cofre = []
    valor = 10.0
  # Act (chamada à função a ser testada)
    cofre_atualizado = adicionar_deposito(cofre, valor)
  # Assert (Verificação da saída)
    assert cofre_atualizado == [10.0], "O cofrinho deve ter exatamente 10 reias após o depósito"

def test_adicionar_deposito_em_cofre_com_valores_existentes():

  cofre = [100,250]
  valor = 50
  cofre_atualizado = adicionar_deposito(cofre, valor)
  assert cofre_atualizado == [100,250,50], "Os depósitos anteriores não podem ser perdidos ao adicionar um novo"

def test_adicionar_deposito_de_valor_zero():

  cofre = [100]
  valor = 0
  cofre_atualizado = adicionar_deposito(cofre, valor)
  assert cofre_atualizado == [100], "A função não aceita aceita"


def test_calcular_total_com_cofre_vazio():

  cofre = []
  total = calcular_total_economizado(cofre)
  assert total == 0, "O cofre está vazio"

def test_calcular_total_com_um_unico_deposito():

  cofre = [100]
  total = calcular_total_economizado(cofre)
  assert total == 100, "O total deve ser o valor do depósito"

def test_calcular_total_com_multiplos_depositos():

  cofre = [100,250,300]
  total = calcular_total_economizado(cofre)
  assert total == 650, "O total deve ser a soma dos depósitos"

def test_meta_nao_atingida_com_total_abaixo():

  total = 500
  meta = 1000
  resultado = meta_atingida(total, meta)
  assert resultado is False, "A meta não foi atingida"


def test_meta_atingida_com_total_acima():

  total = 2000
  meta = 1000
  resultado = meta_atingida(total, meta)
  assert resultado is True, "A meta foi atingida porém o valor ficou acima do esperado"


def test_meta_atingida_com_total_exatamente_igual_a_meta():

  total = 1000
  meta = 1000
  resultado = meta_atingida(total, meta)
  assert resultado is True, "Como a regra usa '>=', o total igual à meta deve ser considerado como meta atingida"


















```
