```python
%%writefile sistema.py
def adicionar_item(carrinho, nome, preco, quantidade):
  """Adiciona um item ao carrinho e retorna o carrinho atualizado."""
  item = {"nome": nome, "preco": preco, "quantidade": quantidade}
  carrinho.append(item)
  return carrinho

def calcular_total(carrinho):
  """Retorna o valor total do carrinho (soma de preco * quantidade)."""
  total = 0
  for item in carrinho:
    total += item["preco"] * item["quantidade"]
  return round(total, 2)

def tem_frete_gratis(valor_total):
  """Retorna True se o valor total for >= 150.00."""
  return valor_total >= 150.00
```

```python
%%writefile conftest.py
import pytest
from sistema import adicionar_item, calcular_total, tem_frete_gratis

@pytest.fixture
def carrinho():
  return []

@pytest.fixture
def carrinho_com_um_item(carrinho):
  # Setup: cria o carrinho vazio (da fixture) e adiciona um item
  adicionar_item(carrinho, "caneca", 25.0, 2)
  # Entrega o carrinho criado ao teste
  yield carrinho
  # Teardown: Limpa o carrinho após o teste
  carrinho.clear()


@pytest.fixture
def carrinho_com_multiplos_itens(carrinho):
  adicionar_item(carrinho, "caneca", 25.0, 2)
  adicionar_item(carrinho, "livro", 40.0, 1)
  yield carrinho
  carrinho.clear()

@pytest.fixture
def carrinho_com_frete_gratis(carrinho):
  adicionar_item(carrinho, "caneca", 25.0, 2)
  adicionar_item(carrinho, "livro", 40.0, 1)
  adicionar_item(carrinho, "teclado", 80.0, 1)
  yield carrinho
  carrinho.clear()
```

```python
%%writefile test_sistema.py
import pytest
from sistema import adicionar_item, calcular_total, tem_frete_gratis

@pytest.fixture
def carrinho():
  return []

@pytest.fixture
def carrinho_com_um_item(carrinho):
  # Setup: cria o carrinho vazio (da fixture) e adiciona um item
  adicionar_item(carrinho, "caneca", 25.0, 2)
  # Entrega o carrinho criado ao teste
  yield carrinho
  # Teardown: Limpa o carrinho após o teste
  carrinho.clear()

def test_adicionar_item_adiciona_um_item_ao_carrinho(carrinho_com_um_item):
  assert len(carrinho_com_um_item) == 1
  assert carrinho_com_um_item[0]['nome'] == 'caneca'
  assert carrinho_com_um_item[0]['preco'] == 25.0
  assert carrinho_com_um_item[0]['quantidade'] == 2


@pytest.fixture
def carrinho_com_multiplos_itens(carrinho):
  adicionar_item(carrinho, "caneca", 25.0, 2)
  adicionar_item(carrinho, "livro", 40.0, 1)
  yield carrinho
  carrinho.clear()


def test_adicionar_itens_preserva_itens_anteriores(carrinho_com_multiplos_itens):
  assert len(carrinho_com_multiplos_itens) == 2
  assert calcular_total(carrinho_com_multiplos_itens) == 90.0



```
